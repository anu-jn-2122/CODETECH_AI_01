```python
# READ THE FILE
import pandas as pd

df = pd.read_csv("/content/myfile.zip")
df.head()
```

```python
# CHECKING MISSING VALUES
df.isnull().sum()
```

```python
# DATA CLEANING [TASK-1]
# DROPPING DUPLICATES
df = df.drop_duplicates()
print(df)
```

```python
# CHECKING DUPLICATE ROWS
total_duplicates = df.duplicated().sum()
print("Total duplicate rows:", total_duplicates)
```

```python
# DATA CLEANING
# FILLING ALL THE MISSING VALUES
df = df.fillna({
    'Meets criteria for meaningful use of EHRs': "Not Specified",
    'Hospital overall rating footnote': "Unknown",
    'Mortality national comparison footnote': "Unknown",
    'Safety of care national comparison footnote': "Not Specified",
    'Readmission national comparison footnote': "Not Specified",
    'Patient experience national comparison footnote': "Not Shared",
    'Effectiveness of care national comparison footnote': "Unknown",
    'Timeliness of care national comparison footnote': "Not Disclosed",
    'Efficient use of medical imaging national comparison footnote': "Not Specified"
})

df.head()
```

```python
# MISSING/NaN VALUE DETECTION
print(df.isnull().sum())
```

```python
# GETTING THE QUICK STRUCTURE OF THE DATAFRAME
df.info()
```

```python
# GETTING STATISTICAL SUMMARY OF NUMERIC COLUMNS
print(df.describe())

```

```python
# NUMERICAL COLUMNS
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
print(numeric_cols)
```

```python
# CATEGORICAL COLUMS
categorical_cols = df.select_dtypes(include=['object']).columns
print(categorical_cols)
```

```python
# UNIVARIATE AND MULTIVARIATE EDE [TASK-2]
# UNIVARIATE EDE
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Plot frequency for each categorical column
for col in categorical_cols:
    plt.figure(figsize=(10,5))
    if df[col].nunique() < 30:  # Only plot if categories are not too many
        sns.countplot(y=col, data=df, order=df[col].value_counts().index)
        plt.title(f"{col} distribution")
        plt.show()
    else:
        print(f"{col} has too many unique values ({df[col].nunique()}) — skipping plot")
        print(df[col].value_counts().head(10))  # Show top 10
```

```python
# MULTIVARIATE EDE
# Example 1: Hospital Rating vs Hospital Type

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

plt.figure(figsize=(12,6))
sns.boxplot(x='Hospital Type', y='Hospital overall rating', data=df)
plt.xticks(rotation=45)
plt.title("Hospital Overall Rating by Hospital Type")
plt.show()


```

```python
# MULTIVARIATE EDE
# Example 2: City vs Hospital Type (categorical vs categorical)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

ct = pd.crosstab(df['City'], df['Hospital Type'])
ct.plot(kind='bar', stacked=True, figsize=(12,6))
plt.title("Hospital Type distribution by City")
plt.ylabel("Count")
plt.show()

```

```python
# MULTIVARIATE EDE
# Correlation heatmap for numerical columns (even though they are IDs)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

numeric_cols = ['Provider ID', 'ZIP Code', 'Phone Number']
plt.figure(figsize=(6,4))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation between numeric columns (IDs)")
plt.show()
```

```python
# HOSPITAL INSIGHTS REPORT FOR ADMINISTRATORS [TASK-4]

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# 1. LOAD DATA
     # Setting the file path as '/content/myfile.zip'
     # df = pd.read_csv('myfile.csv')


# 2️. General Info

print("===== General Information =====")
print(df.info())  # Shows columns, data types, missing values
print("\n===== Missing Values =====")
print(df.isnull().sum())  # Shows how many missing values in each column


# 3️. Hospitals by Location

plt.figure(figsize=(12,6))
df['State'].value_counts().plot(kind='bar')
plt.title("Number of Hospitals by State")
plt.xlabel("State")
plt.ylabel("Number of Hospitals")
plt.show()

# Top 10 cities with most hospitals
plt.figure(figsize=(12,6))
df['City'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Cities by Number of Hospitals")
plt.xlabel("City")
plt.ylabel("Number of Hospitals")
plt.xticks(rotation=45)
plt.show()


# 4️. Hospital Type & Ownership

plt.figure(figsize=(8,5))
sns.countplot(y='Hospital Type', data=df)
plt.title("Hospital Types")
plt.show()

plt.figure(figsize=(8,5))
sns.countplot(y='Hospital Ownership', data=df)
plt.title("Hospital Ownership")
plt.show()


# 5️. Hospital Overall Ratings

plt.figure(figsize=(8,5))
sns.histplot(df['Hospital overall rating'].dropna(), kde=True, bins=10)
plt.title("Distribution of Hospital Ratings")
plt.xlabel("Rating")
plt.show()

# Top 5 hospitals by rating
top_hospitals = df[['Hospital Name', 'City', 'State', 'Hospital overall rating']].dropna()
top_hospitals = top_hospitals.sort_values(by='Hospital overall rating', ascending=False).head(5)
print("\n===== Top 5 Hospitals =====")
print(top_hospitals)


# 6️. Emergency Services

plt.figure(figsize=(6,4))
sns.countplot(x='Emergency Services', data=df)
plt.title("Emergency Services Availability")
plt.show()


# 7️. Quality Metrics

quality_cols = [
    'Mortality national comparison',
    'Readmission national comparison',
    'Safety of care national comparison',
    'Patient experience national comparison'
]

for col in quality_cols:
    plt.figure(figsize=(8,4))
    sns.countplot(y=col, data=df)
    plt.title(f"{col} Distribution")
    plt.show()



```
